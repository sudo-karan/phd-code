```python
# --- CELL: Generate & Download Progress Report ---
# 1. Install the library
!pip install python-docx -q

from docx import Document
from docx.shared import Inches, Pt
from docx.enum.text import WD_ALIGN_PARAGRAPH
from google.colab import files
import datetime

def generate_and_download_report():
    print("Generating Word Document...")
    doc = Document()

    # --- Title Page ---
    title = doc.add_heading('Project Progress Report: Geospatial Ecological Clustering', 0)
    title.alignment = WD_ALIGN_PARAGRAPH.CENTER

    doc.add_paragraph(f'Date: {datetime.date.today()}')
    doc.add_paragraph('Region: Sanjay Van, Delhi')
    doc.add_paragraph('Method: Google Earth Engine (Python API) & SNIC Clustering')
    doc.add_paragraph('-' * 70)

    # --- 1. Executive Summary ---
    doc.add_heading('1. Executive Summary', level=1)
    doc.add_paragraph(
        "We have successfully developed a multi-sensor machine learning pipeline to classify "
        "ecological zones within Sanjay Van. The pipeline harmonizes Optical (Sentinel-2), "
        "Radar (Sentinel-1), Lidar (ETH Canopy Height), and Climate data into a unified 10m feature vector."
    )
    doc.add_paragraph(
        "Key Findings: We empirically demonstrated that using derived ecological metrics "
        "(Amplitude/Phase) produces significantly cleaner clusters compared to raw mathematical coefficients. "
        "Furthermore, we applied SNIC (Object-Based) segmentation to eliminate pixel noise."
    )

    # --- 2. Data Pipeline ---
    doc.add_heading('2. Data Fusion Pipeline', level=1)
    doc.add_paragraph("The following datasets were ingested and harmonized:")

    # Create a table for data sources
    table = doc.add_table(rows=1, cols=3)
    table.style = 'Table Grid'
    hdr = table.rows[0].cells
    hdr[0].text = 'Sensor'
    hdr[1].text = 'Feature Type'
    hdr[2].text = 'Derived Metric'

    data = [
        ('Sentinel-2', 'Phenology', 'NDVI Mean, Amplitude, Phase'),
        ('ETH Global Canopy', 'Structure', 'Height (m), Spatial Texture'),
        ('Sentinel-1 SAR', 'Biomass', 'Backscatter Ratio (VH/VV)'),
        ('CHIRPS', 'Climate', 'Rainfall Anomaly'),
        ('VIIRS', 'Human Pressure', 'Nighttime Lights'),
        ('MODIS', 'Disturbance', 'Years Since Fire')
    ]

    for sensor, ftype, metric in data:
        row = table.add_row().cells
        row[0].text = sensor
        row[1].text = ftype
        row[2].text = metric

    # --- 3. Comparison Experiment ---
    doc.add_heading('3. Experiment: Coefficients vs. Metrics', level=1)
    doc.add_paragraph(
        "We tested four clustering scenarios to determine the optimal feature set. "
        "Visual inspection confirmed that Coefficient-based models resulted in 'salt-and-pepper' "
        "noise, while Metric-based models (Amplitude/Phase) produced contiguous ecological zones."
    )

    # --- 4. Cluster Interpretation ---
    doc.add_heading('4. Ecological Cluster Definitions', level=1)
    doc.add_paragraph("Statistical profiling of the final model (Scenario 4) revealed distinct functional types:")

    # Create a table for cluster definitions
    table2 = doc.add_table(rows=1, cols=2)
    table2.style = 'Table Grid'
    hdr2 = table2.rows[0].cells
    hdr2[0].text = 'Observed Traits'
    hdr2[1].text = 'Ecological Label'

    definitions = [
        ('High Canopy Height + Low Seasonality', 'Stable / Evergreen Forest'),
        ('Low Height + High Seasonality', 'Scrub / Seasonal Vegetation'),
        ('High Night Lights + Low NDVI', 'Urban Edge / Encroachment'),
        ('Low Height + Recent Fire', 'Disturbed / Regrowth Zone')
    ]

    for trait, label in definitions:
        row = table2.add_row().cells
        row[0].text = trait
        row[1].text = label

    # --- 5. Technical Resolutions ---
    doc.add_heading('5. Technical Resolutions', level=1)
    p = doc.add_paragraph()
    p.add_run('Issue: "Collection query aborted" (Memory Limit)').bold = True
    p.add_run('\nResolution: Shifted from Client-Side clustering to Server-Side GEE clustering.')

    p = doc.add_paragraph()
    p.add_run('Issue: "Red Box" / Single Cluster in SNIC').bold = True
    p.add_run('\nResolution: Data was un-normalized. Applied Z-Score Standardization before SNIC clustering.')

    # --- Save and Download ---
    filename = 'Ecological_Clustering_Report.docx'
    doc.save(filename)
    print(f"Report saved as: {filename}")

    print("Triggering download...")
    files.download(filename)

# Run it
generate_and_download_report()
```
